### 1.2 Auditoria e Metadados
Para fins de reprodutibilidade, inspecionamos o arquivo de *log* gerado pela coleta. Este arquivo comprova o rastreio da origem, data de execução e volume de dados ingeridos.

```python
# Leitura e exibição formatada do log de metadados
log_path = LOGS_DIR / "coleta_metadata.json"

if log_path.exists():
    with open(log_path, 'r', encoding='utf-8') as f:
        log = json.load(f)
        
        print("-" * 50)
        print("AUDITORIA DE COLETA - METADADOS".center(50))
        print("-" * 50)
        print(f"Projeto: {log['projeto']}")
        print(f"Fonte: {log['fonte_dados']['nome']}")
        print(f"Endpoint: {log['fonte_dados']['url_base']}")
        print("-" * 50)
        print(f"Legislatura: {log['escopo_temporal']['legislatura']}")
        print(f"Total de Registros: {log['resumo']['total_registros']:,}")
        print(f"Data da Execução: {log['data_execucao']}")
        print("-" * 50)
else:
    print("Log de coleta não encontrado. Verifique a execução do script anterior.")
```
